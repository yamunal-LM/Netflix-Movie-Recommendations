```python
with zipfile.ZipFile("netflix_titles.csv.zip", "r") as z:
    z.extractall("netflix_data")

print(os.listdir("netflix_data"))
```

```python
df = pd.read_csv("netflix_data/netflix_titles.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
print(df.head())
```

```python

# Fill missing values
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Not Listed")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Not Rated")

# Remove rows without date_added
df = df.dropna(subset=["date_added"]).copy()

# Convert date_added into datetime
df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    errors="coerce"
)

# Remove invalid dates
df = df.dropna(subset=["date_added"]).copy()

# Extract the year content was added
df["year_added"] = df["date_added"].dt.year

print("Data cleaning completed!")
print("Remaining rows:", len(df))
print(df[["title", "country", "rating",
          "date_added", "year_added"]].head())
```

```python

# OPTION 1: Count Movies vs TV Shows
def count_types():
    print("\nMovies vs TV Shows")
    print(df["type"].value_counts())


# OPTION 2: Number of titles added each year
def titles_per_year():
    print("\nTitles Added Each Year")
    print(df["year_added"].value_counts().sort_index())


# OPTION 3: Top 10 producing countries
def top_countries():
    countries = df["country"].str.split(", ").explode()
    countries = countries[countries != "Unknown"]

    print("\nTop 10 Countries")
    print(countries.value_counts().head(10))


# OPTION 4: Top 5 most common genres
def top_genres():
    genres = df["listed_in"].dropna().str.split(", ").explode()

    print("\nTop 5 Genres")
    print(genres.value_counts().head(5))


# OPTION 5: Rating distribution
def rating_distribution():
    print("\nRating Distribution")
    print(df["rating"].value_counts())


# OPTION 6: Oldest and newest release years
def release_years():
    years = df["release_year"].dropna()

    print("\nOldest Release Year:", int(years.min()))
    print("Newest Release Year:", int(years.max()))


# OPTION 7: Search titles by keyword
def search_titles():
    keyword = input("Enter movie/show keyword: ").strip()

    if not keyword:
        print("Please enter a keyword.")
        return

    result = df[df["title"].str.contains(
        keyword, case=False, na=False
    )]

    if result.empty:
        print("No titles found.")
    else:
        print(result[["title", "type", "release_year",
                      "rating"]].to_string(index=False))


# OPTION 8: Content added in a specific year
def content_by_year():
    try:
        year = int(input("Enter year (e.g., 2019): "))
        result = df[df["year_added"] == year]

        print("\nTitles added in", year)
        if result.empty:
            print("No content found for this year.")
        else:
            print(result[["title", "type",
                          "year_added"]].to_string(index=False))
            print("Total:", len(result))

    except ValueError:
        print("Please enter a valid year.")


# OPTION 9: Total unique countries
def unique_countries():
    countries = df["country"].str.split(", ").explode()
    countries = countries[countries != "Unknown"].str.strip()

    print("\nTotal unique countries:",
          countries.nunique())


# OPTION 10: Summary statistics using NumPy
def year_statistics():
    years = df["release_year"].dropna().to_numpy()

    print("\nRelease Year Statistics")
    print("Mean:", round(np.mean(years), 2))
    print("Median:", np.median(years))
    print("Minimum:", np.min(years))
    print("Maximum:", np.max(years))
    print("Standard Deviation:", round(np.std(years), 2))


print("All 10 functions are ready!")
```

```python
while True:
    print("\n===== NETFLIX CONTENT EXPLORER =====")
    print("1. Count Movies vs TV Shows")
    print("2. Number of titles added each year")
    print("3. Top 10 producing countries")
    print("4. Top 5 most common genres")
    print("5. Rating distribution")
    print("6. Oldest and newest release years")
    print("7. Search titles by keyword")
    print("8. Content added in a specific year")
    print("9. Total unique countries")
    print("10. Summary statistics using NumPy")
    print("11. Quit")

    choice = input("Enter your choice (1-11): ").strip()

    if choice == "1":
        count_types()

    elif choice == "2":
        titles_per_year()

    elif choice == "3":
        top_countries()

    elif choice == "4":
        top_genres()

    elif choice == "5":
        rating_distribution()

    elif choice == "6":
        release_years()

    elif choice == "7":
        search_titles()

    elif choice == "8":
        content_by_year()

    elif choice == "9":
        unique_countries()

    elif choice == "10":
        year_statistics()

    elif choice == "11":
        print("Thank you for using Netflix Content Explorer!")
        break

    else:
        print("Invalid choice. Enter a number from 1 to 11.")
```

```python
# Fix incorrect duration values in rating column
mask = df["rating"].str.contains(
    r"\bmin\b", case=False, na=False
)

duration_missing = mask & df["duration"].isna()

df.loc[duration_missing, "duration"] = df.loc[
    duration_missing, "rating"
]

df.loc[mask, "rating"] = "Not Rated"

print("Rating data cleaned successfully!")
print(df["rating"].value_counts())
```
